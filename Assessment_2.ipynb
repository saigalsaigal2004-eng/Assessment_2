```python
filepath = '/content/drive/MyDrive/DSAB13/pandas assesment/assesment_pandas/student_performance.csv'
student_df = pd.read_csv(filepath)
print(student_df)
```

```python
#Display the first 5 records.
#Display the last 5 records.
#Find the number of rows and columns.
#Display the column names.
#Display the data types.
#Generate a statistical summary of the numerical columns.
print(student_df.head())
print(student_df.tail())
print(student_df.shape)
print(student_df.columns)
print(student_df.dtypes)
print(student_df.describe)
```

```python
#Find students whose Final_Score is greater than 75.
#Find students whose Attendance is below 75%.
#Find students from the Data Science department with Final_Score greater than 80.
#Sort all students by Final_Score from highest to lowest.
print(student_df[student_df["Final_Score"] > 75])
print(student_df[student_df["Attendance"] < 75])
print(student_df[(student_df["Department"] == "Data Science") & (student_df["Final_Score"] > 80)])
print(student_df.sort_values(by="Final_Score",ascending=False))
```

```python
student_df['Average_Technical_Score'] = (student_df['Python_Score'] + student_df['SQL_Score'] + student_df['Pandas_Score'] + student_df['NumPy_Score']) / 4
print(student_df)
```

```python
def get_performance(score):
    if score >= 90:
        return "Excellent"
    elif score >= 75:
        return "Good"
    elif score >= 60:
        return "Average"
    else:
        return "Needs Improvement"

student_df["Performance"] = student_df[
    "Average_Technical_Score"
].apply(get_performance)
print(student_df)
```

```python
#Number of students
#Average Final Score
#Maximum Final Score
#Minimum Final Score
#Sort the resulting DataFrame by Average Final Score in descending order
print("The Number of students:",student_df.shape[0])
print("Average Final Score:",student_df["Final_Score"].mean())
print("Maximum Final Score:",student_df["Final_Score"].max())
print("Minimum Final Score:",student_df["Final_Score"].min())
print(student_df.sort_values(by="Final_Score",ascending=False))
```

```python
scores = student_df["Final_Score"].to_numpy()


mean_score = np.mean(scores)
print("Mean:", mean_score)
print("Median:", np.median(scores))


print("Standard Deviation:", np.std(scores))


print("25th Percentile:", np.percentile(scores, 25))

print("75th Percentile:", np.percentile(scores, 75))


above_mean = student_df[student_df["Final_Score"] > mean_score]

print("\nStudents above the overall mean:")
print(above_mean)
```

```python
filepath = '/content/drive/MyDrive/DSAB13/pandas assesment/assesment_pandas/student_performance.csv'
student_df = pd.read_csv(filepath)
print(student_df)
```

```python
#Identify missing values.
#Handle missing numerical values appropriately.
print(student_df.isnull().sum())
numeric_columns = [
    "Age",
    "Attendance",
    "Python_Score",
    "SQL_Score",
    "Pandas_Score",
    "NumPy_Score",
    "Final_Score"
]

for col in numeric_columns:
    student_df[col] = student_df[col].fillna(student_df[col].mean())
```

```python
student_df["Average_Score"] = student_df[
    [
        "Python_Score",
        "SQL_Score",
        "Pandas_Score",
        "NumPy_Score"
    ]
].mean(axis=1)
print(student_df)
```

```python
student_df["Pass_Status"] = np.where(
    student_df["Average_Score"] >= 50,
    "Pass",
    "Fail"
)
print(student_df)
```

```python
#Top 5 students based on Average_Score .
#Department having the highest average score.
#Average attendance by department.
#Number of students who passed and failed.
print(student_df.sort_values(by="Average_Score",ascending=False).head(5))
print(student_df.groupby("Department")["Average_Score"].mean())
print(student_df.groupby("Department")["Attendance"].mean())
print(student_df["Pass_Status"].value_counts())
```
